# Minggu 1: Praktikum 2 - Visualisasi Distribusi


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../datasetPBL/Transactions.csv')

# 1. Distribusi NetSales
plt.figure(figsize=(10,6))
sns.histplot(df['NetSales'], kde=True, color='blue')
plt.title('Distribusi NetSales')
plt.show()
```

```python
# 2. Frekuensi Outlet
plt.figure(figsize=(10,6))
df['Outlet'].value_counts().plot(kind='bar', color='green')
plt.title('Frekuensi Transaksi per Outlet')
plt.xlabel('Outlet')
plt.ylabel('Jumlah Transaksi')
plt.show()
```
