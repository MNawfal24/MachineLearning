# Minggu 6: Praktikum 3 - Vector Search Sederhana


```python
import pandas as pd
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler

df = pd.read_csv('../datasetPBL/Transactions.csv').sample(5000, random_state=42).reset_index(drop=True)
X = df[['NetSales', 'TotalAmount']]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

nn = NearestNeighbors(n_neighbors=5, algorithm='auto')
nn.fit(X_scaled)

# Cari transaksi yang mirip dengan baris pertama
distances, indices = nn.kneighbors([X_scaled[0]])

print("Indeks Transaksi Terdekat:", indices)
```
